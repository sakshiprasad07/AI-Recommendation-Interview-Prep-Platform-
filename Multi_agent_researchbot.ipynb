```python
import os
os.makedirs("config", exist_ok=True)

agents_yaml = """
research_planner:
  role: >
    Research Planner
  goal: >
    Analyze queries and break them down into smaller, specific research topics.
  backstory: >
    You are a research strategist who excels at breaking down complex questions
    into manageable research components. You identify what needs to be researched
    and create clear research objectives.

internet_researcher:
  role: >
    Internet Researcher
  goal: >
    Research thoroughly all assigned topics
  backstory: >
    You are a meticulous online researcher who knows how to find reliable,
    up-to-date information quickly. You use search and scraping tools to
    gather detailed data and always keep track of where information came from.

fact_checker:
  role: >
    Fact Checker
  goal: >
    Verify data for accuracy, identify inconsistencies, and flag potential misinformation
  backstory: >
    You are a rigorous fact-checker with a sharp eye for inconsistencies,
    outdated claims, and unreliable sources. You cross-reference information
    across multiple sources before confirming it as accurate.

report_writer:
  role: >
    Report Writer
  goal: >
    Write clear, concise, and well-structured reports based on gathered information
  backstory: >
    You are a skilled technical writer who transforms raw research and verified
    facts into polished, well-organized reports that are easy for anyone to read
    and act on.
"""

tasks_yaml = """
create_research_plan:
  description: >
    Based on the user's query, break it down into specific topics and key questions,
    and create a focused research plan. The user's query is: {user_query}
  expected_output: >
    A research plan with main research topics to investigate, key questions for
    each topic, and success criteria for the research.

gather_research_data:
  description: >
    "Using the research plan, collect information on all identified topics. "
        "When searching, always use specific, natural-language search phrases "
        "(e.g. 'AI video generation 2026 trends') — never pass generic category "
        "labels like 'research papers' or 'industry reports' as a search query. "
        "Cite all sources used."
  expected_output: >
    A collection of research findings for each topic in the research plan,
    including key facts, data points, and a list of cited source URLs for each finding.

verify_information_quality:
  description: >
    Review all collected research. Identify any conflicting information,
    potential misinformation, or gaps that need addressing.
  expected_output: >
    A verification report listing confirmed facts, flagged inconsistencies or
    questionable claims, identified gaps in the research, and recommendations
    for what still needs clarification.

write_final_report:
  description: >
    Create a comprehensive report that answers the original query using all
    verified research data. Structure it with clear sections, include citations,
    and provide actionable insights.
  expected_output: >
    A polished, well-structured final report in markdown format, with clear
    headings, a Summary section, an Insights section, a Citations section listing
    all sources, and a concluding section of actionable insights that directly
    answers the original user query.
"""

with open("config/agents.yaml", "w") as f:
    f.write(agents_yaml)

with open("config/tasks.yaml", "w") as f:
    f.write(tasks_yaml)

print("✓ config/agents.yaml and config/tasks.yaml created")
```

```python
from crewai import Agent, Task, Crew, LLM
import os
os.environ["CREWAI_TESTING"] = "true"

from google.colab import userdata
groq_api_key = userdata.get('GROQ_API_KEY').strip()

groq_llm = LLM(
    model="llama-3.1-8b-instant",
    provider="openai",
    base_url="https://api.groq.com/openai/v1",
    api_key=groq_api_key,
    max_tokens=1024,
    temperature=0.3,
)

# define the research planner agent
research_planner = Agent(
    role="Research Planner",
    goal="Analyze queries and break them down into smaller, specific research topics.",
    backstory=(
         "You are a research strategist who excels at breaking down complex questions "
         "into manageable research components. You identify what needs to be researched "
         "and create clear research objectives."
    ),
    llm=groq_llm,
    verbose=True,
    max_rpm=150,
    max_iter=15
)

researcher = Agent(
    role="Internet Researcher",
    goal="Research thoroughly all assigned topics",
    backstory=(
        "You are a knowledgeable researcher who synthesizes what you know about "
        "each topic clearly and thoroughly, drawing on your training knowledge "
        "to provide detailed, accurate information with proper attribution where possible."
    ),
    # tools removed — no more search/scrape tool-call context buildup
    llm=groq_llm,
    verbose=True,
    max_rpm=150,
    max_iter=15
)

fact_checker = Agent(
    role="Fact Checker",
    goal=(
        "Verify data for accuracy, identify inconsistencies, "
        "and flag potential misinformation"
    ),
    backstory=(
        "You are a rigorous fact-checker with a sharp eye for inconsistencies, "
        "outdated claims, and unreliable sources. You review claims against your "
        "own knowledge and flag anything questionable or unverifiable."
    ),
    llm=groq_llm,
    verbose=True,
    max_rpm=150,
    max_iter=15
)

report_writer = Agent(
    role="Report Writer",
    goal="Write clear, concise, and well-structured reports based on gathered information",
    backstory=(
         "You are a skilled technical writer who transforms research and verified "
         "facts into polished, well-organized reports that are easy for anyone to read "
         "and act on. You always fully complete every section you start."
    ),
    llm=groq_llm,
    verbose=True,
    max_rpm=150,
    max_iter=15
)

# define the create research plan task
create_research_plan_task = Task(
    description=(
        "Based on the user's query, break it down into specific topics and key questions, "
        "and create a focused research plan."
        "The user's query is: {user_query}"
    ),
    expected_output=(
        "A research plan with main research topics to investigate, "
        "key questions for each topic, and success criteria for the research."
        ),
    agent=research_planner,
)

# define the gather research data task
gather_research_data_task = Task(
    description=(
        "Using the research plan, collect information on all identified topics "
        "based on your own knowledge. Cite general sources where relevant."
    ),
    expected_output=(
        "A collection of research findings for each topic in the research plan, "
        "including key facts and data points."
    ),
    agent=researcher
)

# define the verify information quality task
verify_information_quality_task = Task(
    description=(
        "Review all collected research. Identify any conflicting information, "
        "potential misinformation, or gaps that need addressing."
    ),
    expected_output=(
        "A verification report listing confirmed facts, flagged inconsistencies or "
        "questionable claims, identified gaps in the research, and recommendations "
        "for what still needs clarification."
    ),
    agent=fact_checker
)

# define the write final report task
write_final_report_task = Task(
    description=(
        "Create a comprehensive report that answers the original query using all verified research data. "
        "Structure it with clear sections, include citations, and provide actionable insights."
    ),
    expected_output=(
        "A polished, well-structured final report in markdown format, with clear headings, "
        "cited sources, and a concluding section of actionable insights that directly answers "
        "the original user query."
    ),
    agent=report_writer
)

# create the crew with the defined agents and tasks
crew = Crew(
    agents=[research_planner, researcher, fact_checker, report_writer],
    tasks=[create_research_plan_task, gather_research_data_task, verify_information_quality_task, write_final_report_task]
)

user_query = "What are the latest trends in AI agents for content creation in 2026?"

result = await crew.kickoff_async(
    inputs={
        "user_query": user_query,
    }
)

from IPython.display import Markdown
Markdown(result.raw)
```
